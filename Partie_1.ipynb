```python
def infoOut(data,description=None,details=False):
    dfInfo = data.columns.to_frame(name='Column')
    dfInfo['Null Count'] = data.isnull().sum()
    dfInfo['Dtype'] = data.dtypes

    dfInfo['min']=data.iloc[:,0:-1].min()
    dfInfo['max']=data.iloc[:,0:-1].max()
    
    if description:
        dfInfo['Description'] = description
    dfInfo.reset_index(drop=True,inplace=True)
    if details:
        rangeIndex = (dfInfo['Null Count'].min(),dfInfo['Non-Null Count'].min())
        totalColumns = dfInfo['Column'].count()
        dtypesCount = dfInfo['Dtype'].value_counts()
        totalMemory = dfInfo.memory_usage().sum()
        return dfInfo, rangeIndex, totalColumns, dtypesCount, totalMemory
    else:
        return dfInfo
```

### Prétraitement 

#### Choix de la méthode de remplacement des valeurs manquantes.


#####  Use the attribute median for all samples belonging to the same class as the given tuple

```python
dataset_1.info()
```

```python
data = dataset_1.groupby(['Fertility']).transform(lambda x: x.fillna(x.median()))
```

```python
data = pd.concat([dataset_1['Fertility'], data], axis=1)
```

```python
data
```

####  Choix de la méthode de traitement des valeurs aberrantes

```python
##### Visulaiser les outliers par catégorie de fertilité
for column in numeric_columns:
    fig = px.box(data, y=column, x='Fertility')
    fig.update_layout(width=800, height=600,title=f"Boxplot for {column} Attribute")
    fig.show()
    fig.write_image(f"Box_categ_{column}.png")
```

##### Utilisation IQR 

```python
# supprimer le valeurs aberrantes  val < lower_bound or val > upper_bound de chaque catégorie de fertilité
new_dataset_1=data
for column in data.columns[1:]:
    for fertility in data['Fertility'].unique():
        data_per_class = data[data['Fertility']==fertility][column]
        IQR = data_per_class.quantile(0.75) - data_per_class.quantile(0.25)
        r = 1.5 * IQR
        lower_bound = data_per_class.quantile(0.25) - r
        upper_bound = data_per_class.quantile(0.75) + r
        new_dataset_1 = new_dataset_1[((new_dataset_1['Fertility'] != fertility) | ((new_dataset_1['Fertility'] == fertility) & (new_dataset_1[column] > lower_bound) & (new_dataset_1[column] <= upper_bound)))]
```
